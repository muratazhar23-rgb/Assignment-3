```python
def bayes(prior, observation):
    if observation == "ALARM":
        p_h = 0.90
        p_s = 0.10
    else:
        p_h = 0.10
        p_s = 0.90

    total = p_h * prior + p_s * (1 - prior)
    return p_h * prior / total


print("ALARM:", round(bayes(0.20, "ALARM"), 4))
print("NORMAL:", round(bayes(0.20, "NORMAL"), 4))
```

```python
priors = [0.05, 0.20, 0.50, 0.80]

for p in priors:
    result = bayes(p, "ALARM")
    print(p, round(result, 4))
```

```python
observations = ["ALARM", "NORMAL", "ALARM", "ALARM"]

p_h = 0.20

for i, obs in enumerate(observations):
    if i == 0:
        predicted_h = p_h
    else:
        predicted_h = 0.80 * p_h + 0.10 * (1 - p_h)

    predicted_s = 1 - predicted_h

    p_h = bayes(predicted_h, obs)
    p_s = 1 - p_h

    print(
        "t" + str(i + 1),
        obs,
        "P(H) =", round(p_h, 4),
        "P(S) =", round(p_s, 4)
    )
```
